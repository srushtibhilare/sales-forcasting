```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    r'C:\Users\srush\OneDrive\Desktop\AI-Sales-Forecasting-Demand-Planning\data\raw_sales.csv'
)
```

```python
df.head()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(' ', '_')
      .str.replace('-', '_')
)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='%d-%m-%Y')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='%d-%m-%Y')
```

```python
df['postal_code'] = df['postal_code'].astype('Int64').astype(str)
```

```python
df['sales'] = df['sales'].round(2)
```

```python
df = df.drop_duplicates()
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_month_name'] = df['order_date'].dt.month_name()
```

```python
df['delivery_days'] = (df['ship_date'] - df['order_date']).dt.days
```

```python
df['sales_category'] = pd.cut(
    df['sales'],
    bins=[0, 100, 500, 2000, df['sales'].max()],
    labels=['Low', 'Medium', 'High', 'Very High']
)
```

```python
df.info()
df.head()
```

```python
monthly_sales = (
    df
    .groupby(['order_year', 'order_month'], as_index=False)
    .agg(
        total_sales=('sales', 'sum'),
        total_orders=('order_id', 'nunique'),
        avg_delivery_days=('delivery_days', 'mean')
    )
)
```

```python
monthly_sales['date'] = pd.to_datetime(
    monthly_sales['order_year'].astype(str) + '-' +
    monthly_sales['order_month'].astype(str) + '-01'
)

monthly_sales = monthly_sales.sort_values('date')
monthly_sales.head()
```

```python
train = monthly_sales[monthly_sales['date'] < '2017-01-01']
test  = monthly_sales[monthly_sales['date'] >= '2017-01-01']
```

```python
test['baseline_forecast'] = train['total_sales'].mean()
```

```python
from sklearn.linear_model import LinearRegression

X_train = train[['order_year', 'order_month']]
y_train = train['total_sales']

X_test = test[['order_year', 'order_month']]
y_test = test['total_sales']

model = LinearRegression()
model.fit(X_train, y_train)

test['lr_forecast'] = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, test['lr_forecast'])
print("MAE:", mae)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(train['date'], train['total_sales'], label='Train')
plt.plot(test['date'], test['total_sales'], label='Actual')
plt.plot(test['date'], test['lr_forecast'], label='Forecast')
plt.legend()
plt.title('Monthly Sales Forecast')
plt.show()
```
