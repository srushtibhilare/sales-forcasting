```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.holtwinters import ExponentialSmoothing
```

```python
monthly_sales = pd.read_csv(
    r'C:\Users\srush\OneDrive\Desktop\AI-Sales-Forecasting-Demand-Planning\data\monthly_sales.csv',
    parse_dates=['date']
)

monthly_sales = monthly_sales.sort_values('date')
monthly_sales.head()
```

```python
train_size = int(len(monthly_sales) * 0.8)

train = monthly_sales.iloc[:train_size]
test = monthly_sales.iloc[train_size:]
```

```python
model = ExponentialSmoothing(
    train['total_sales'],
    trend='add',
    seasonal='add',
    seasonal_periods=12
)

hw_model = model.fit()
```

```python
predictions = hw_model.forecast(len(test))
```

```python
mae = mean_absolute_error(test['total_sales'], predictions)
rmse = np.sqrt(mean_squared_error(test['total_sales'], predictions))

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
```

```python
forecast_df = pd.DataFrame({
    'date': test['date'].values,
    'actual_sales': test['total_sales'].values,
    'predicted_sales': predictions.values
})

forecast_df.head()
```

```python
forecast_df.to_csv(
    r'C:\Users\srush\OneDrive\Desktop\AI-Sales-Forecasting-Demand-Planning\data\forecast_results.csv',
    index=False
)
```
